```python
import pandas as pd

df = pd.read_excel('/content/CapstoneProject-Beyond-Predictive-Accuracy/Diabetes Simple Diagnosis.csv.xlsx')
df.head()
```

```python
# Shape: rows and columns
print("Shape:", df.shape)
```

```python
# 2. Column names and data types
df.info()
```

```python
# 3. Missing (null) values
print(df.isnull().sum())
```

```python
# 4. Duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# 5. Basic statistics
df.describe()
```

```python
# Drop the useless index column
df = df.drop(columns=['Unnamed: 0'])

# Gender: describe() skipped it
print(df['Gender'].value_counts())

# Suspicious ages
print("Age == 0:", (df['Age'] == 0).sum())
print("Age < 5:", (df['Age'] < 5).sum())

# Suspicious BMI
print("BMI < 15:", (df['BMI'] < 15).sum())
print("BMI > 60:", (df['BMI'] > 60).sum())

# Target balance
print(df['Diagnosis'].value_counts())
print(df['Diagnosis'].value_counts(normalize=True) * 100)

# Does FBS and HbA1c separate the two classes?
print(df.groupby('Diagnosis')[['FBS', 'HbA1c_level', 'Age', 'BMI']].mean())
```
